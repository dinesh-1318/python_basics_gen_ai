## **if else statement**
### **question** :largest of three numbers get input from user


```python
a=int(input("enter the 1 number:"))
b=int(input("enter the 2 number:"))
c=int(input("enter the 3 number:"))
if a>b and a>c:
    print("a is bigger number")
elif b>a and b>c:
    print("b is bigger number")
else:
    print("c is bigger number")
```
