```python

import json

data = {"name": "Arun", "age": 22,"city":("tamil nadu","chennai"), "year": None}

# Dictionary to JSON string
s = json.dumps(data)
print(s, type(s))

# JSON string to dictionary
d = json.loads(s)
print(d, type(d))

# Write to JSON file
with open("../arun.json", "w") as f:
    json.dump(data, f)

# Read from JSON file
with open("../arun.json", "r") as f:
    result = json.load(f)

print(result)
```
