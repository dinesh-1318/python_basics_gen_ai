## looping
### while loop 
**question**-print the below pattern:
```text
12345
12345
12345
12345
12345
```

````python
a=1
rows= int(input("enter the no of rows:"))
columns= int(input("enter the no of columns:"))
while a <= rows:
    b=1
    while b <= columns:
        print(b,end='')
        b += 1
    a+=1
    print()
````

## while loop pattern
**question-print the below pattern and get input from user**
```text
###
###
###
```

````python
rows= int(input("enter the no of rows:"))
columns= int(input("enter the no of columns:"))
symbol=input("enter the symbol:")
a=1
while a <= rows:
    b=1
    while b <= columns:
        print(symbol,end='')
        b += 1
    a+=1
    print()
````

# for loop
*1.question*
```text
****
****
****
****
```

````python
rows= int(input("enter the no of rows:"))
columns= int(input("enter the no of columns:"))
symbol=input("enter the symbol:")
for i in range(rows):
    for j in range(columns):
        print(symbol,end='')
    print()
````

````python
rows= int(input("enter the no of rows:"))
symbol=input("enter the symbol:")
for i in range(rows):
    for j in range(0,i+1):
        print(symbol,end='')
    print()
````
