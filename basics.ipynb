# **Part B** *– Operators*

```python
#  4.odd or even
n = int(input())
if n % 2 == 0:
    print("Even")
else:
    print("Odd")
```

```python
# 6.arithmetic operator
a = 15
b = 4
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponent:", a ** b)
```

```python
# 7.positive or negative
n = int(input())

if n > 0:
    print("Positive")
elif n < 0:
    print("Negative")
else:
    print("Zero")
```
