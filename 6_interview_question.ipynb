### **1.reverse words** 

```python
def reverse_words(mystr):       # This is a string
    words = mystr.split()       # ["This","is","a","string"]
    reverse = words[::-1]       # ["string","is","a","This"]
    result = ' '.join(reverse)  # string is a This
    print(result)
reverse_words("This is a string")
```

### **2.palindrome**

```python
def is_palindrome(mystr):
    newstr = ""
    for each in mystr:
        if mystr.isalpha():
            newstr = newstr + each.lower()
    if newstr == newstr[::-1]:
        return True
    else:
        return False
is_palindrome("Markram")
```

### **3.count frequency**

```python
def count_freq(mystr):
    newstr = ""
    mydict = {}
    for i in mystr:
        if i.isalpha():
            newstr = newstr + i.lower()
    for j in newstr:
        if j not in mydict.keys():
            mydict[j] = 1
        else:
            mydict[j] += 1
    print(mydict)
count_freq("this is dinesh writing code")
```

### **4.Remove duplicate character**

```python
def remove_duplicate(mystr):
    seen = set()
    result=[]
    for i in mystr:
        if i.isalpha():
            if i not in seen:
                seen.add(i)
                result.append(i)
    print(result)
remove_duplicate("this is a is string")
```

### **5.Remove duplicate word**

```python
def remove_duplicate2(mystr):
    seen = set()
    result = []
    for i in mystr.split():
        if i not in seen:
            seen.add(i)
            result.append(i)
    print(" ".join(result))
remove_duplicate2("this is dinesh and this code was write by him")
```
